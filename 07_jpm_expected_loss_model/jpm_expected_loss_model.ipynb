```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score , confusion_matrix
import numpy as np


def expected_loss_function(
    credit_lines_outstanding,
    loan_amt_outstanding,
    total_debt_outstanding,
    income,
    years_employed,
    fico_score,
    recovery_rate=0.10
):
    borrower_df = pd.DataFrame({
        "credit_lines_outstanding": [credit_lines_outstanding],
        "loan_amt_outstanding": [loan_amt_outstanding],
        "total_debt_outstanding": [total_debt_outstanding],
        "income": [income],
        "years_employed": [years_employed],
        "fico_score": [fico_score]
    })

    probability_default = model.predict_proba(borrower_df)[0][1]

    loss_given_default = 1 - recovery_rate

    expected_loss = probability_default * loan_amt_outstanding * loss_given_default

    return {
        "Probability_Default": probability_default,
        "Probability_Default_Percentage": probability_default * 100,
        "Expected_Loss": expected_loss
    }

np.set_printoptions(suppress=True, precision=6)
pd.set_option("display.float_format", "{:.6f}".format)

df = pd.read_csv('Task 3 and 4_Loan_Data.csv')

feature_df = ['credit_lines_outstanding','loan_amt_outstanding','total_debt_outstanding','income','years_employed','fico_score']

split_df = int(len(df)*0.7)
train_df = df.iloc[:split_df]
test_df = df.iloc[split_df:]

X_train_df = train_df[feature_df]
y_train_df = train_df['default']

X_test_df = test_df[feature_df]
y_test_df = test_df['default']

model = LogisticRegression(max_iter=1000)

model.fit(X_train_df,y_train_df)

result = expected_loss_function(
    credit_lines_outstanding=3,
    loan_amt_outstanding=15000,
    total_debt_outstanding=25000,
    income=60000,
    years_employed=5,
    fico_score=650
)

print(result,'result')

model_predict = model.predict(X_test_df)


accuracy_score(y_test_df,model_predict)
confusion_matrix(y_test_df,model_predict)
probability = model.predict_proba(X_test_df)
print(probability)

default_probability_df = pd.DataFrame({
    'Real_Default' : y_test_df.values ,
    'Probability_Default' : probability[:,1],
    'Probability_No_Default' : probability[:,0],
    'Probability_Default_Percentage' : probability[:,1] * 100
})
X_test_df_reset = X_test_df.reset_index(drop = True)
X_test_df_reset
final_df = pd.concat([X_test_df_reset,default_probability_df],axis=1)
final_df

final_df["Loss_Given_Default_Amount"] = final_df["loan_amt_outstanding"] * 0.90
final_df["Expected_Loss"] = final_df["Probability_Default"] * final_df["loan_amt_outstanding"] * 0.90

```

This model predicts the probability of default using borrower features such as credit lines outstanding, loan amount outstanding, total debt outstanding, income, years employed, and FICO score. Since default is a binary outcome, I used Logistic Regression. The model outputs the probability of default using predict_proba. Expected loss is calculated as Probability of Default × Exposure at Default × Loss Given Default. Since the recovery rate is 10%, the loss given default is 90%.
